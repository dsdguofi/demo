```python
positive_samples = train_df[train_df["Heart Disease"] == "Presence"]
negative_samples = train_df[train_df["Heart Disease"] == "Absence"]
```

```python
median_positive_age = positive_samples["Age"].median()
median_negative_age = negative_samples["Age"].median()
print(f'Median Age for Presence of Heart Attack: {median_positive_age}')
print(f'Median Age for Absence of Heart Attack: {median_negative_age}')
```
